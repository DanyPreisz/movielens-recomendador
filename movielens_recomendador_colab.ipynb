# Filtrado Colaborativo de Películas (MovieLens)

Proyecto de sistema de recomendación listo para **Google Colab**.

**Pregunta:** ¿cómo recomendar películas basándonos en ratings de usuarios?

**Método:** Matrix Factorization (SVD) con Surprise

**Stack:** pandas, surprise, matplotlib

Dataset MovieLens 100k.

## 0. Setup

```python
!pip install -q scikit-surprise

import pandas as pd
import matplotlib.pyplot as plt
from surprise import Dataset, Reader, SVD
from surprise.model_selection import train_test_split, cross_validate
print("OK")
```

## 1. Cargar MovieLens 100k

```python
!wget -q https://files.grouplens.org/datasets/movielens/ml-100k.zip
!unzip -q ml-100k.zip

df = pd.read_csv("ml-100k/u.data", sep="\t", names=["user", "item", "rating", "timestamp"])
print(df.shape)
df.head()
```

## 2. EDA

```python
print(df["rating"].describe())
plt.hist(df["rating"], bins=5)
plt.title("Distribución de ratings")
plt.show()
```

## 3. Preparar datos para Surprise

```python
reader = Reader(rating_scale=(1, 5))
data = Dataset.load_from_df(df[["user", "item", "rating"]], reader)
trainset, testset = train_test_split(data, test_size=0.2, random_state=42)
print("Datos preparados")
```

## 4. Modelo SVD

```python
algo = SVD(random_state=42)
algo.fit(trainset)

predictions = algo.test(testset)
print("Modelo entrenado")
```

## 5. Evaluación

```python
from surprise import accuracy
print(f"RMSE: {accuracy.rmse(predictions):.4f}")
print(f"MAE: {accuracy.mae(predictions):.4f}")
```

## 6. Recomendaciones de ejemplo

```python
# Predecir rating para un usuario y película
pred = algo.predict(uid=196, iid=302)
print(f"Predicción para usuario 196, película 302: {pred.est:.2f}")
```

## 7. Conclusiones

- SVD es un método clásico y efectivo para filtrado colaborativo.
- MovieLens es el dataset de referencia para recomendadores.
